```python
import osmnx as ox
import numpy as np
place_name = "Khartoum, Sudan"
# Define your target year and format it for Overpass
year = 2021
date_str = f"{year}-01-01T00:00:00Z"

# Ensure the overpass endpoint supports attic data
ox.settings.overpass_endpoint = "https://overpass-api.de/api"

# Inject the date into overpass settings
ox.settings.overpass_settings = f"[out:json][date:'{date_str}']{{maxsize}}"

# Fetch the graph (example coordinates: downtown Raleigh, NC)
#point = (35.7796, -78.6382)
lulc_tags = {
    "landuse": True,
    "amenity": True,
    "building": True,
    "natural": True
    #"natural": ["wood", "water", "wetland", "scrub", "grassland", "bare_rock", "sand"],
    #"leisure": ["park", "nature_reserve"]
    #"amenity": "school"
}

print(f"Fetching polygons for {place_name}...")
lat, lon = 49.2128, 37.2569
# Extract the geometries
# Note: In newer versions of osmnx, features_from_place replaces geometries_from_place
bbox = ox.utils_geo.bbox_from_point(point=(lat, lon), dist=20000)#ox.features_from_place(place_name, tags=lulc_tags)
gdf = ox.features_from_bbox(bbox=bbox, tags=lulc_tags)

gdf['building'] = gdf['building'].replace('yes', np.nan)
target_cols = ['landuse', 'natural', 'building', 'amenity', 'military', 'healthcare', 'shop']
target_cols = [col for col in target_cols if col in gdf.columns]

gdf = gdf[gdf.geom_type.isin(["Polygon", "MultiPolygon"])]
gdf = gdf.dropna(subset=target_cols, how = 'all')
gdf['feature'] = gdf[target_cols].bfill(axis=1).iloc[:, 0]
#gdf.explore(
#    column='natural',  # The column determining the colors
#    cmap='Set3',       # A qualitative colormap ideal for categories
#    tooltip='natural', # What shows up when you hover over a polygon
#    popup=True,        # Clicking shows all attributes
#    tiles="CartoDB positron" # Clean, light background map
#)
```
