```python
import matplotlib.pyplot as plt
def get_rgb_image(image):
    img = image.sel(band=[3, 2, 1])
    print(img.shape)
    b04 = img[0, :, :]  # Red band
    b03 = img[1, :, :]  # Green band
    b02 = img[2, :, :]  # Blue band
    
    # Stack into RGB image
    rgb_image = (np.stack([b04, b03, b02], axis=-1) * 255).astype(np.uint8)
    print(rgb_image.shape)
    return rgb_image

plt.figure(figsize=(8, 8))
rgb_image = get_rgb_image(image2)
#img.sel(band=[3, 2, 1]).plot.imshow(robust=True)
plt.title("Sentinel Image")

min_val, max_val = np.percentile(rgb_image, (2, 98)) # Optional: Use percentiles
scaled_band = np.clip(rgb_image, min_val, max_val)
rgb_image = (rgb_image - min_val) / (max_val - min_val)

plt.imshow(rgb_image)
plt.show()
```
