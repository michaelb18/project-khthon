```python
%matplotlib inline
def display_rgb_image(image_tensor, ax=None):
    """
    Display B02, B03, B04 bands as RGB image.
    
    Bands are in order: B1, B2, B3, B4, B5, B6, B7, B8A, B8B, B9, B10, B11, B12
    B4 (index 3) = Red, B3 (index 2) = Green, B2 (index 1) = Blue
    """
    # Convert tensor to numpy if needed
    image = image_tensor
    
    # Extract B02 (index 1), B03 (index 2), B04 (index 3)
    # B4=Red, B3=Green, B2=Blue
    b04 = image[3, :, :]  # Red band
    b03 = image[2, :, :]  # Green band
    b02 = image[1, :, :]  # Blue band
    
    # Stack into RGB image
    rgb_image = np.stack([b04, b03, b02], axis=-1)
    
    # Normalize to [0, 1] range for display
    # Sentinel-2 values are typically in reflectance range [0, 1] or [0, 10000]
    # Apply percentile-based normalization to handle outliers
    #for i in range(3):
    #    band = rgb_image[:, :, i]
    #    p2, p98 = np.percentile(band, [2, 98])
    #    rgb_image[:, :, i] = np.clip((band - p2) / (p98 - p2 + 1e-8), 0, 1)
    
    plt.imshow(rgb_image)
    plt.axis('off')
    
    #return ax
import cv2
import numpy as np
import matplotlib.pyplot as plt

def draw_text_and_box(image_rgb, top_left, bottom_right, text, font_scale=1, font_color=(255, 0, 0), thickness=2):
    
    # Draw the bounding box (rectangle)
    cv2.rectangle(image_rgb, top_left, bottom_right, (0, 255, 0), thickness)
    
    # Choose a font
    font = cv2.FONT_HERSHEY_SIMPLEX
    
    # Calculate the position for the text (a little inside the box)
    text_size = cv2.getTextSize(text, font, font_scale, thickness)[0]
    text_x = top_left[0] + 10
    text_y = top_left[1] + (text_size[1] + 10)
    
    # Draw the text on the image
    cv2.putText(image_rgb, text, (text_x, text_y), font, font_scale, font_color, thickness, cv2.LINE_AA)
    
    return image_rgb

def rgb_histogram_equalization(img):
    # 1. Convert the image from BGR to YCbCr color space
    # OpenCV loads images in BGR format by default
    ycrcb_img = cv2.cvtColor(img, cv2.COLOR_BGR2YCR_CB)
    
    # 2. Split the image into Y, Cb, and Cr channels
    channels = cv2.split(ycrcb_img)
    y_channel = channels[0]
    
    # 3. Apply histogram equalization to the Y (luminance) channel
    cv2.equalizeHist(y_channel, y_channel)
    
    # 4. Merge the equalized Y channel back with the original Cb and Cr channels
    cv2.merge(channels, ycrcb_img)
    
    # 5. Convert the image back to BGR (RGB) color space
    equalized_img = cv2.cvtColor(ycrcb_img, cv2.COLOR_YCR_CB2BGR)
    
    return equalized_img
```
