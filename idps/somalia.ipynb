```python
import pandas as pd

somalia = pd.read_excel('250217_cccmcluster_masterlist.xlsx')
somalia['Settlement Type'] = somalia['Settlement Type'].str.lower()
somalia = somalia[somalia['Settlement Type'] == 'planned camp/site']
```

```python
for lat, lon, households in zip(somalia['Lat'], somalia['Long'], somalia['Latest HHs']):
    print(lat, lon)
    image2 = get_image(lat, lon, 3000, 3000, days = 50)
    img = turn_into_image(image2)
    plt.imshow(img)
    plt.title(f'Households = {households}')
    plt.show()
```

```python
img = np.array(image2.sel(band=[3, 2, 1]))
img = np.moveaxis(img, 0, -1)
min_val, max_val = np.percentile(img, (2, 98)) # Optional: Use percentiles
scaled_band = np.clip(img, min_val, max_val)
rgb_image = (scaled_band - min_val) / (max_val - min_val)
plt.imshow(rgb_image)
```
